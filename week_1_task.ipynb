SWAPPING BETWEEN TWO NUMBERS

```python
# Task 1: Swap two numbers without third variable
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(f"Before swap: a = {a}, b = {b}")

# Approach: Python tuple unpacking
a, b = b, a

print(f"After swap: a = {a}, b = {b}")
```

CHECK PRIME AND PALENDROME

```python
# Task 2: Check if number is prime AND if string is palindrome

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def is_palindrome(s):
    s = s.lower().replace(" ", "")  # ignore case and spaces
    return s == s[::-1]

num = int(input("Enter a number: "))
text = input("Enter a string: ")

print(f"{num} is Prime: {is_prime(num)}")
print(f"'{text}' is Palindrome: {is_palindrome(text)}")
```

 Fibonacci - Loop + Recursion

```python
# Task 3: Fibonacci series using loop and recursion

def fib_loop(n):
    a, b = 0, 1
    print("Fibonacci using Loop:", end=" ")
    for _ in range(n):
        print(a, end=" ")
        a, b = b, a + b
    print()

def fib_recursion(n):
    if n <= 1:
        return n
    return fib_recursion(n-1) + fib_recursion(n-2)

n = int(input("Enter how many terms: "))
fib_loop(n)

print("Fibonacci using Recursion:", end=" ")
for i in range(n):
    print(fib_recursion(i), end=" ")
```

Function Library

```python
# Task 4: Temperature converter + Area calculator

import math

def temp_converter(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

def area_circle(radius):
    return math.pi * radius**2

def area_rectangle(length, width):
    return length * width

c = float(input("Enter temperature in Celsius: "))
print(f"{c}C = {temp_converter(c)}F")

r = float(input("Enter radius of circle: "))
print(f"Area of Circle: {area_circle(r):.2f}")

l = float(input("Enter length: "))
w = float(input("Enter width: "))
print(f"Area of Rectangle: {area_rectangle(l, w)}")
```

Inheritance - Shape, Circle, Rectangle

```python
# Task 5: 3 classes demonstrating inheritance

import math

class Shape:
    def area(self):
        pass  # to be overridden

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius**2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width

c = Circle(5)
r = Rectangle(4, 6)
print(f"Circle Area: {c.area():.2f}")
print(f"Rectangle Area: {r.area()}")
```

Exception-safe input

```python
# Task 6: Keep asking until valid integer is entered

while True:
    try:
        num = int(input("Enter a valid integer: "))
        print(f"You entered: {num}")
        break  # exit loop if input is valid
    except ValueError:
        print("Invalid input! Please enter an integer.")
```

Mini Project: Student Grade Calculator

```python
# Mini Project: Command-line Student Grade Calculator

def get_grade(marks):
    if marks >= 90: return 'A'
    elif marks >= 80: return 'B'
    elif marks >= 70: return 'C'
    elif marks >= 60: return 'D'
    else: return 'F'

def get_valid_marks():
    while True:
        try:
            marks = float(input("Enter marks 0-100: "))
            if 0 <= marks <= 100:
                return marks
            else:
                print("Marks must be between 0 and 100")
        except ValueError:
            print("Invalid input! Enter a number.")

name = input("Enter student name: ")
marks = get_valid_marks()
grade = get_grade(marks)

print(f"\nStudent: {name}")
print(f"Marks: {marks}")
print(f"Grade: {grade}")
```
