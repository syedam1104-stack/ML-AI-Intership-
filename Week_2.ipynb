Task 1: Read/Write CSV and JSON

```python
import csv
import json

# Write CSV
data = [["Name", "Age", "City"], ["Ali", 20, "Karachi"], ["Sara", 22, "Lahore"]]
with open("data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(data)

# Read CSV and filter
with open("data.csv", "r") as f:
    reader = csv.DictReader(f)
    filtered = [row for row in reader if int(row["Age"]) > 20]
print("Filtered CSV:", filtered)

# Write JSON
json_data = {"students": [{"name": "Ali", "age": 20}]}
with open("data.json", "w") as f:
    json.dump(json_data, f, indent=4)

# Read JSON
with open("data.json", "r") as f:
    loaded = json.load(f)
print("Loaded JSON:", loaded)
```

Task 3: Pandas - Load, Filter, Group, Handle Missing

```python
import pandas as pd

# Create sample data
df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Sara"],
    "City": ["Karachi", "Lahore", "Karachi", "Lahore"],
    "Marks": [85, None, 90, 78]
})

print("Original Data:\n", df)

# Filter rows
filtered = df[df["Marks"] > 80]
print("\nFiltered >80:\n", filtered)

# Group by City
grouped = df.groupby("City")["Marks"].mean()
print("\nAverage Marks by City:\n", grouped)

# Handle missing values
df["Marks"].fillna(df["Marks"].mean(), inplace=True)
print("\nAfter filling missing:\n", df)
```

Task 4: Merge 2 datasets

```python
import pandas as pd

df1 = pd.DataFrame({"ID": [1, 2], "Name": ["Ali", "Sara"]})
df2 = pd.DataFrame({"ID": [1, 2], "Marks": [85, 90]})

merged = pd.merge(df1, df2, on="ID")
print("Merged Data:\n", merged)
print("\nSummary:\n", merged.describe())
```

Task 5: Custom Exception + Logging

```python
import logging

logging.basicConfig(filename="app.log", level=logging.ERROR, format="%(asctime)s - %(levelname)s - %(message)s")

class NegativeValueError(Exception):
    pass

def check_value(x):
    try:
        if x < 0:
            raise NegativeValueError("Value cannot be negative")
    except NegativeValueError as e:
        logging.error(e)
        print("Error logged")

check_value(-5)
```

Mini Project: Data Cleaning Tool

```python
import pandas as pd

# Load messy CSV
df = pd.read_csv("messy_data (1).csv")

report = {}
report["duplicates_removed"] = df.duplicated().sum()
df.drop_duplicates(inplace=True)

report["missing_before"] = df.isnull().sum().sum()
df.fillna(df.mean(numeric_only=True), inplace=True)

# Fix inconsistent formats - example: lowercase city names
df["Name"] = df["NAME"].str.lower().str.title() # Corrected column name
df["CITY"] = df["CITY"].str.lower().str.title() # Corrected column name

df.to_csv("cleaned_data.csv", index=False)

print("Cleaning Report:")
for k, v in report.items():
    print(f"{k}: {v}")
print("Cleaned file saved as cleaned_data.csv")
```
