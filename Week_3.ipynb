Task 1: Bar, Line, Pie Chart with Matplotlib

```python
import matplotlib.pyplot as plt

data = {"Jan": 20, "Feb": 35, "Mar": 30, "Apr": 45}
months = list(data.keys())
values = list(data.values())

# Bar Chart
plt.bar(months, values, color="skyblue")
plt.title("Bar Chart - Monthly Sales")
plt.show()

# Line Chart
plt.plot(months, values, marker="o", color="green")
plt.title("Line Chart - Monthly Sales")
plt.show()

# Pie Chart
plt.pie(values, labels=months, autopct="%1.1f%%")
plt.title("Pie Chart - Monthly Sales")
plt.show()
```

Task 2: Correlation Heatmap with Seaborn

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Math": [85, 90, 78, 92],
    "Science": [88, 85, 80, 95],
    "English": [75, 88, 82, 90]
})

corr = df.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

Task 3: Fetch data from Public REST API using requests

```python
import requests
import json

url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    with open("api_data.json", "w") as f:
        json.dump(data, f, indent=4)
    print("Data saved to api_data.json")
else:
    print("Failed to fetch data")
```

Task 4: Parse JSON API and plot part of it

```python
import requests
import matplotlib.pyplot as plt

url = "https://jsonplaceholder.typicode.com/users"
data = requests.get(url).json()

# Count how many users per city
cities = [user["address"]["city"] for user in data]
city_counts = {city: cities.count(city) for city in set(cities)}

plt.bar(city_counts.keys(), city_counts.values())
plt.xticks(rotation=45)
plt.title("Users per City from API")
plt.show()
```

Task 5: 2-3 Reusable Plotting Functions

```python
import matplotlib.pyplot as plt

def plot_bar(x, y, title):
    plt.bar(x, y)
    plt.title(title)
    plt.show()

def plot_line(x, y, title):
    plt.plot(x, y, marker="o")
    plt.title(title)
    plt.show()

def plot_pie(labels, sizes, title):
    plt.pie(sizes, labels=labels, autopct="%1.1f%%")
    plt.title(title)
    plt.show()

# Usage
plot_bar(["A", "B", "C"], [10, 20, 15], "My Bar Chart")
```

Mini Project: EDA Mini-Dashboard Script

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load public dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

print("Dataset Shape:", df.shape)
print(df.head())

# Chart 1: Bar - Average tip by day
plt.figure(figsize=(6,4))
df.groupby("day")["tip"].mean().plot(kind="bar")
plt.title("Avg Tip by Day")
plt.show()

# Chart 2: Line - Total bill trend
plt.figure(figsize=(6,4))
df["total_bill"].plot(kind="line")
plt.title("Total Bill Trend")
plt.show()

# Chart 3: Pie - Gender distribution
df["sex"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Gender Distribution")
plt.show()

# Chart 4: Heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="YlGnBu")
plt.title("Correlation Heatmap")
plt.show()

# Insights
print("\nInsights:")
print("1. Tips are highest on Sunday.")
print("2. Total bill has some high-value outliers.")
print("3. Male and female customers are almost equal.")
print("4. Total bill and tip are positively correlated.")
```
